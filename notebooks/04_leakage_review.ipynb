```python
sample = df[
    (df["station_id"] == 117)
    & (df["param_code"] == "PM10")
][
    [
        "date",
        "mean_value",
        "pm_lag_1d",
        "pm_mean_3d",
        "pm_mean_7d",
    ]
].sort_values("date")

sample.head(15)
```

```python
row_index = 4

previous_3 = sample.iloc[
    row_index - 3 : row_index
]["mean_value"]

print("Previous 3 values:")
print(previous_3)

print("\nManual mean:")
print(previous_3.mean())

print("\nFeature pm_mean_3d:")
print(sample.iloc[row_index]["pm_mean_3d"])
```
